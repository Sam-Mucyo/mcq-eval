```python
valid_df = df_train[df_train['at_labels'].notna() & df_train['question_type'].notna()].copy()

valid_df['at_labels'] = valid_df['at_labels'].astype(bool)
valid_df['label_name'] = valid_df['at_labels'].map({
    True: 'Answerable',
    False: 'Unanswerable'
})

question_order = ['What', 'Why', 'When', 'Where', 'Who', 'How', 'None of the above']

count_crosstab = pd.crosstab(
    valid_df['question_type'],
    valid_df['label_name']
).reindex(index=question_order, fill_value=0).reindex(columns=['Answerable', 'Unanswerable'], fill_value=0)

percent_crosstab = count_crosstab.div(count_crosstab.sum(axis=1), axis=0).multiply(100).round(1)

ax = count_crosstab.plot(kind='bar', figsize=(9, 5), color=['skyblue', 'lightcoral'])

for i, label in enumerate(count_crosstab.columns):
    for j, value in enumerate(count_crosstab[label]):
        pct = percent_crosstab.iloc[j, i]
        if value > 0:
            bar = ax.containers[i][j]
            ax.annotate(f'{pct:.1f}%',
                        xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
                        xytext=(0, 2),
                        textcoords='offset points',
                        ha='center', va='bottom', fontsize=9)


plt.title('Answerable vs. Unanswerable by Question Type')
plt.xlabel('Question Type')
plt.ylabel('Number of Questions')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from difflib import SequenceMatcher

# Function to compute similarity ratio between two strings
def similarity_ratio(a, b):
    return SequenceMatcher(None, str(a), str(b)).ratio()

valid_df = df_train[
    df_train['question_AUT_PT_1'].notna() &
    df_train['question_AUT_PT_2'].notna() &
    df_train['answer_AUT_PT_1'].notna() &
    df_train['answer_AUT_PT_2'].notna()
].copy()

valid_df['question_similarity'] = valid_df.apply(
    lambda row: similarity_ratio(row['question_AUT_PT_1'], row['question_AUT_PT_2']), axis=1)

valid_df['answer_similarity'] = valid_df.apply(
    lambda row: similarity_ratio(row['answer_AUT_PT_1'], row['answer_AUT_PT_2']), axis=1)

# Agreement if similarity ≥ 0.70
question_agree = valid_df['question_similarity'] >= 0.70
answer_agree = valid_df['answer_similarity'] >= 0.70

# Count agreements and disagreements
agreement_counts = pd.DataFrame({
    'Agreed': [question_agree.sum(), answer_agree.sum()],
    'Disagreed': [(~question_agree).sum(), (~answer_agree).sum()]
}, index=['Question Annotation (PT)', 'Answer Annotation (PT)'])


ax = agreement_counts.plot(kind='bar', stacked=True, figsize=(8, 5), color=['skyblue', 'lightcoral'])

for col_idx, col_label in enumerate(agreement_counts.columns):
    for row_idx, row_label in enumerate(agreement_counts.index):
        bar = ax.containers[col_idx][row_idx]
        height = bar.get_height()
        base = bar.get_y()
        if height > 0:
            total = agreement_counts.loc[row_label].sum()
            pct = height / total * 100
            ax.annotate(f'{pct:.1f}%',
                        xy=(bar.get_x() + bar.get_width() / 2, base + height / 2),
                        xytext=(0, 0),
                        textcoords='offset points',
                        ha='center', va='center',
                        color='black',
                        fontsize=9)


plt.title('Portuguese Annotator Agreement (Text Similarity ≥ 70%)')
plt.ylabel('Number of Samples')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
